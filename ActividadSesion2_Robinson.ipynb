## **Ejercicio 1: Clasificador de notas**
Crea una función que reciba una nota de 0 a 5 y devuelva el estado del estudiante. Luego aplícala a una lista de notas.

### **Pasos**

```python
# Nota < 3.0 → "Reprueba"; de 3.0 a 4.4 → "Aprueba"; ≥ 4.5 → "Excelente".
def clasificar_nota(nota):
    if nota < 3.0:
        return "Reprueba"
    elif nota < 4.5:
        return "Aprueba"
    else:
        return "Excelente"
```

```python
# Valida: si la nota no está entre 0 y 5, devuelve "Nota inválida".
def clasificar_nota(nota):
    if nota < 0 or nota > 5:
        return "Nota inválida"
    elif nota < 3.0:
        return "Reprueba"
    elif nota < 4.5:
        return "Aprueba"
    else:
        return "Excelente"
```

```python
# Aplica la función con un for a [2.5, 3.8, 4.7, 5.2, 3.0].
notas = [2.5, 3.8, 4.7, 5.2, 3.0]

for nota in notas:
    print(f"Nota", nota, ":", clasificar_nota(nota))
```

```python
# Cuenta cuántos estudiantes aprobaron.
aprobados = 0
for nota in notas:
    if clasificar_nota(nota) in ("Aprueba", "Excelente"):
        aprobados += 1

print(f"Aprobaron", aprobados, "de", len(notas), "estudiantes")
```

## **Ejercicio 2: Carrito de compras**
Una tienda online guarda el carrito como un diccionario producto → (precio, cantidad). Calcula el total de la compra.

### **Pasos**

```python
# Carrito: camiseta (45.000 × 2), jean (120.000 × 1), medias (12.000 × 3).
carrito = {
    "camiseta": (45000, 2),
    "jean": (120000, 1),
    "medias": (12000, 3),
}
```

```python
# Recorre el diccionario y calcula el subtotal.
subtotal = 0
for producto, (precio, cantidad) in carrito.items():
    valor = precio * cantidad
    subtotal += valor
    print(producto, precio, "x", cantidad, "=", valor)

print("Subtotal:", subtotal)
```

```python
# Si el subtotal supera 200.000, aplica 10% de descuento.
if subtotal > 200000:
    descuento = subtotal * 0.10
else:
    descuento = 0

base = subtotal - descuento
print("Descuento:", descuento)
print("Subtotal con descuento:", base)
```

```python
# Suma IVA del 19% e imprime el total con formato de pesos.
def pesos(valor):
    return f"${valor:,.0f}".replace(",", ".")

iva = base * 0.19
total = base + iva

print("Subtotal:", pesos(subtotal))
print("Descuento (10%): -", pesos(descuento))
print("IVA (19%):", pesos(iva))
print("TOTAL A PAGAR:", pesos(total))
```
