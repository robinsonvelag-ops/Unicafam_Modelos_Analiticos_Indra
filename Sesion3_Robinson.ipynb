**Problema: ¿te alcanza para el café?**

```python
# Variables
presupuesto = 100000
precio_cafe = 2500
cafes_semana = 5
```

```python
# 1. ¿Cúanto gastas al mes? (4 semana)
gasto_mes = (precio_cafe * cafes_semana) * 4
print(gasto_mes)
```

```python
# 2. ¿Cúanto te sobra?
sobra = presupuesto - gasto_mes
print(sobra)
```

```python
# 3. ¿Te alcanza?
if sobra >= 0:
  alcanza = True
else
  alcanza = False

print(alcanza)
```
