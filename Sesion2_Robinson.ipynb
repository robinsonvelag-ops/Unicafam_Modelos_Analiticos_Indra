**Taller Big Data**

```python
# Instalación de bibliotecas
import pandas as pd
import numpy as np
```

```python
# Numeros aleatorios
np.random.seed(42)
n = 200
```

```python
# Tabla sintetica de ventas - creación de la tabla
ventas = pd.DataFrame({
    "fecha": pd.date_range("2026-01-01", periods=n, freq="D"),
    "ciudad": np.random.choice(["Bogotá", "Medellin", "Cali", "Barranquilla"], n),
    "canal": np.random.choice(["Tienda", "Web", "App"], n),
    "unidades": np.random.randint(1, 20, n),
    "precio": np.random.choice([25000, 48000, 99000], n),
})
```

```python
# Creacion de la columna total
ventas["total"] = ventas["unidades"] * ventas["precio"]
```

```python
# Ver resultados
print(ventas.shape)
print(ventas.head(3))
```

```python
# Descripcion de tabla
ventas.info()
```

**Analítica Descriptiva ¿Que pasó?**

```python
# Tabla dinamica
resumen = (ventas.groupby("ciudad") ["total"]
           .agg(["count", "sum", "mean"])
           .sort_values("sum", ascending=False))
print(resumen)
```

```python
# Valor en millones
resumen["sum"] = (resumen["sum"] / 1e6).round(1)
resumen["mean"] = resumen["mean"].round(0)
resumen.columns = ["Ventas", "Total_Millones", 'ticket_prom']
```

```python
print(resumen)
```

**Diagnostico ¿Por que pasó?**

```python
tabla = pd.pivot_table(ventas, values="total", index="ciudad",
                       columns="canal", aggfunc="sum") / 1e6
print(tabla.round(1))
print()
print("Canal con más ventas:", ventas.groupby("canal") ["total"].sum().idxmax())
```
